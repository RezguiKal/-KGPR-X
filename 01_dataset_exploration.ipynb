```python
import json
import os
import pandas as pd
from collections import Counter

# ─── CHEMINS ───────────────────────────────────────────────────────────────
BASE_DIR        = r"D:\Memoire\mooccubeX"
TRANSLATED_DIR  = os.path.join(BASE_DIR, "translated")

def path(filename, translated=False):
    """Retourne le chemin complet d'un fichier."""
    folder = TRANSLATED_DIR if translated else BASE_DIR
    return os.path.join(folder, filename)

print("✅ Chemins configurés")
print(f"   Base     : {BASE_DIR}")
print(f"   Traduits : {TRANSLATED_DIR}")
```

```python
def load_json_sample(filepath, n=3, encoding='utf-8'):
    """Charge les n premiers enregistrements d'un fichier JSON Lines ou JSON."""
    records = []
    try:
        with open(filepath, 'r', encoding=encoding, errors='replace') as f:
            # Essai JSON Lines (une entrée par ligne)
            for i, line in enumerate(f):
                if i >= n:
                    break
                line = line.strip()
                if line:
                    try:
                        records.append(json.loads(line))
                    except json.JSONDecodeError:
                        pass
        if records:
            return records, "jsonl"
        # Sinon essai JSON standard
        with open(filepath, 'r', encoding=encoding, errors='replace') as f:
            data = json.load(f)
        if isinstance(data, list):
            return data[:n], "json_list"
        elif isinstance(data, dict):
            items = list(data.items())[:n]
            return items, "json_dict"
    except Exception as e:
        return [], f"ERREUR: {e}"


def count_json_lines(filepath, encoding='utf-8'):
    """Compte le nombre de lignes non-vides dans un fichier."""
    count = 0
    try:
        with open(filepath, 'r', encoding=encoding, errors='replace') as f:
            for line in f:
                if line.strip():
                    count += 1
    except Exception as e:
        return f"ERREUR: {e}"
    return count


def load_txt_sample(filepath, n=5, sep='\t', encoding='utf-8'):
    """Charge les n premières lignes d'un fichier texte (relations)."""
    rows = []
    try:
        with open(filepath, 'r', encoding=encoding, errors='replace') as f:
            for i, line in enumerate(f):
                if i >= n:
                    break
                parts = line.strip().split(sep)
                rows.append(parts)
    except Exception as e:
        return [], f"ERREUR: {e}"
    return rows, "txt"


def file_size_mb(filepath):
    """Retourne la taille d'un fichier en MB."""
    try:
        return round(os.path.getsize(filepath) / (1024 * 1024), 1)
    except:
        return "N/A"


print("✅ Fonctions utilitaires chargées")
```

```python
files_info = [
    # (label, filepath, type)
    ("Courses (traduit)",    path("course_translated.json", translated=True),  "node"),
    ("Videos (traduit)",     path("video_translated.json",  translated=True),  "node"),
    ("Concepts (traduit)",   path("concept_translated.json",translated=True),  "node"),
    ("Teachers (traduit)",   path("teacher_translated.json",translated=True),  "node"),
    ("Schools (traduit)",    path("school_translated.json", translated=True),  "node"),
    ("Users",                path("user.json"),                                "node"),
    ("user-video",           path("user-video.json"),                          "relation"),
    ("video_id-ccid",        path("video_id-ccid.txt"),                        "relation"),
    ("concept-video",        path("concept-video.txt"),                        "relation"),
    ("concept-course",       path("concept-course.txt"),                       "relation"),
    ("course-teacher",       path("course-teacher.txt"),                       "relation"),
    ("course-school",        path("course-school.txt"),                        "relation"),
]

rows = []
for label, fpath, ftype in files_info:
    exists = os.path.exists(fpath)
    size   = file_size_mb(fpath) if exists else "MANQUANT"
    lines  = count_json_lines(fpath) if exists else "MANQUANT"
    rows.append({"Fichier": label, "Type": ftype, "Taille (MB)": size, "Lignes": lines, "Existe": "✅" if exists else "❌"})

df_summary = pd.DataFrame(rows)
print("\n📊 Récapitulatif des fichiers du KG :\n")
print(df_summary.to_string(index=False))
```

```python
# ── Charger un échantillon de course IDs ──────────────────────────────────
course_ids = set()
with open(path("course_translated.json", translated=True), 'r', encoding='utf-8', errors='replace') as f:
    for i, line in enumerate(f):
        if i >= 5000:
            break
        line = line.strip()
        if line:
            try:
                obj = json.loads(line)
                cid = obj.get("id") or obj.get("course_id") or obj.get("_id")
                if cid:
                    course_ids.add(str(cid))
            except:
                pass

print(f"✅ Cours chargés (échantillon) : {len(course_ids)}")
print(f"   Exemples d'IDs : {list(course_ids)[:5]}")
```

```python
# ── Vérifier concept-course.txt ───────────────────────────────────────────
matched = 0
total   = 0
sample_pairs = []

with open(path("concept-course.txt"), 'r', encoding='utf-8', errors='replace') as f:
    for i, line in enumerate(f):
        if i >= 10000:
            break
        parts = line.strip().split('\t')
        if len(parts) >= 2:
            total += 1
            course_in_rel = str(parts[1]).strip()
            if course_in_rel in course_ids:
                matched += 1
            if i < 5:
                sample_pairs.append(parts)

print(f"\n📌 concept-course.txt — vérification sur {total} lignes :")
print(f"   IDs cours reconnus : {matched} / {total} ({100*matched//total if total else 0}%)")
print(f"   Séparateur utilisé : tab")
print(f"   Exemples de paires :")
for p in sample_pairs:
    print(f"     {p}")
```

```python
# ── Charger concept-course pour analyser la co-occurrence ─────────────────
from collections import defaultdict

concept_to_courses = defaultdict(set)
course_to_concepts = defaultdict(list)  # liste ordonnée si possible

with open(path("concept-course.txt"), 'r', encoding='utf-8', errors='replace') as f:
    for line in f:
        parts = line.strip().split('\t')
        if len(parts) >= 2:
            concept_id = parts[0].strip()
            course_id  = parts[1].strip()
            concept_to_courses[concept_id].add(course_id)
            course_to_concepts[course_id].append(concept_id)

print(f"✅ Concepts distincts      : {len(concept_to_courses):,}")
print(f"✅ Cours avec concepts     : {len(course_to_concepts):,}")

# Statistiques
concepts_per_course = [len(v) for v in course_to_concepts.values()]
print(f"\n📊 Concepts par cours :")
print(f"   Min   : {min(concepts_per_course)}")
print(f"   Max   : {max(concepts_per_course)}")
print(f"   Moy   : {sum(concepts_per_course)/len(concepts_per_course):.1f}")

# Exemple d'un cours et ses concepts
ex_course = list(course_to_concepts.keys())[0]
print(f"\n📄 Exemple : cours '{ex_course}' contient {len(course_to_concepts[ex_course])} concepts")
print(f"   Premiers concepts : {course_to_concepts[ex_course][:5]}")
```

```python
# ── Paires concept-concept co-occurrentes (mini-test sur 500 cours) ────────
from itertools import combinations

co_occurrence = Counter()
sample_courses = list(course_to_concepts.items())[:500]

for course_id, concepts in sample_courses:
    # Paires ordonnées (position i avant position j → prérequis potentiel)
    for i in range(len(concepts)):
        for j in range(i+1, min(i+4, len(concepts))):  # fenêtre de 3
            co_occurrence[(concepts[i], concepts[j])] += 1

print(f"✅ Paires concept-concept générées (sur 500 cours) : {len(co_occurrence):,}")

# Top paires
top_pairs = co_occurrence.most_common(10)
print("\n🔝 Top 10 paires co-occurrentes (concept_A → prérequis de → concept_B) :")
for (c1, c2), cnt in top_pairs:
    print(f"   {c1} → {c2}  (co-occurrence : {cnt})")
```
